```python
from bs4 import BeautifulSoup
import cloudscraper

scraper = cloudscraper.create_scraper(browser={'browser': 'chrome', 'platform': 'windows', 'desktop': True})
r = scraper.get("https://www.magicsur.cl/15-juegos-de-mesa-magicsur-chile?page=1", timeout=45)
soup = BeautifulSoup(r.text, 'html.parser')

# Check what selectors actually exist
print("Product containers found:", len(soup.find_all(class_='product-miniature')))
print("Product titles found:", len(soup.find_all(class_='product-title')))

# Dump first product block to inspect actual structure
first_product = soup.find(class_='product-miniature')
if first_product:
    print("\nFirst product raw HTML:")
    print(first_product.prettify()[:2000])
else:
    print("No products at all. Site structure differs.")
```
